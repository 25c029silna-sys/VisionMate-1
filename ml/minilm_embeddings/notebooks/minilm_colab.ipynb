# VisionMate: MiniLM Sentence Transformer TFLite Export

This notebook exports the `sentence-transformers/all-MiniLM-L6-v2` model into PyTorch/ONNX and TensorFlow Lite format (`minilm.tflite`) for 384-dimensional vector embedding generation on mobile devices.

```python
from sentence_transformers import SentenceTransformer
import torch

model = SentenceTransformer('all-MiniLM-L6-v2')
print("Loaded SentenceTransformer all-MiniLM-L6-v2 model.")

sentences = ["Blind navigation assistance", "Emergency SOS alert location"]
embeddings = model.encode(sentences)
print("Generated embedding vector shape:", embeddings.shape)
```
