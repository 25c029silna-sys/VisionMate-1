# VisionMate: YOLOv8 Obstacle Detection & TFLite Export

This notebook fine-tunes Ultralytics YOLOv8 Nano on indoor obstacle categories and exports the trained weights into TensorFlow Lite (`yolov8n.tflite`) for real-time mobile scene navigation.

```python
!pip install ultralytics
```

```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')
print("Loaded YOLOv8 Nano model.")

# Export to TFLite format
exported_path = model.export(format='tflite')
print(f"TFLite model exported successfully to: {exported_path}")
```
